# Step 2 — unpack `sw_pynq_upload.zip` on the board

Run this notebook **once** after you upload `sw_pynq_upload.zip` through Jupyter **Files → Upload**.

It looks for the zip in the notebook directory, your home folder, and the current working directory, then extracts it **next to the zip** (you get a `sw/` folder).

```python
import zipfile
from pathlib import Path

here = Path.cwd().resolve()
home = Path.home().resolve()
candidates = [
    here / "sw_pynq_upload.zip",
    home / "sw_pynq_upload.zip",
    (here.parent / "sw_pynq_upload.zip"),
]

zip_path = next((p for p in candidates if p.is_file()), None)
if zip_path is None:
    raise FileNotFoundError(
        "Could not find sw_pynq_upload.zip. Tried:\n  " + "\n  ".join(str(p) for p in candidates)
    )

dest = zip_path.parent
print(f"Found: {zip_path}")
print(f"Extracting into: {dest}")

with zipfile.ZipFile(zip_path, "r") as zf:
    zf.extractall(dest)

print("Done. Open Files → sw → board_a_dashboard → Board_A_Dashboard.ipynb")
```

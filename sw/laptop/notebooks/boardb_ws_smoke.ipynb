# TradeMark Board B — WebSocket smoke test\n\nRun `react_boardb_server.py` first, then run the cell below to connect to the same `/ws/stream` feed the React UI uses.

```python
import json\nimport time\n\nimport websocket\n\nWS = "ws://127.0.0.1:8765/ws/stream"\n\n# If you run Vite dev on :5173 and want to hit that origin instead:\n# WS = "ws://127.0.0.1:5173/ws/stream"\n\nws = websocket.create_connection(WS, timeout=2)\n\n# Tell server what you want (optional; matches the React UI prefs)\nws.send(json.dumps({"history_sec": 120, "ema_symbol": 0}))\n\nprint("connected ->", WS)\n\nfor i in range(6):\n    raw = ws.recv()\n    pkt = json.loads(raw)\n    ts = pkt.get("ts")\n    age = pkt.get("age_ms")\n    hb = pkt.get("heartbeat")\n    connected = pkt.get("connected")\n    symbols = (pkt.get("meta") or {}).get("symbols")\n    latest = pkt.get("latest") or {}\n\n    print(f"\npacket {i}:")\n    print("  ts:", ts)\n    print("  connected:", connected, "heartbeat:", hb, "age_ms:", age)\n    if symbols:\n        print("  symbols[0:5]:", symbols[:5])\n    print("  latest keys:", sorted(list(latest.keys()))[:18], "...")\n\n    time.sleep(0.15)\n\nws.close()\nprint("\nclosed")\n
```
